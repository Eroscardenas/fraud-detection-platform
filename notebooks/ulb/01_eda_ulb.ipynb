# Credit Card Fraud Detection — Exploratory Data Analysis

This notebook explores the ULB credit card transaction dataset before building the fraud detection model.

### Objectives
- Understand the dataset
- Check data quality
- Analyze class imbalance
- Explore transaction amounts
- Identify initial fraud patterns

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import plotly.express as px
```

```python
# Cargar dataset
df = pd.read_csv("../../data/raw/ulb/creditcard.csv")

# Ver tamaño del dataset
print(df.shape)

# Ver primeras filas
df.head()
```

```python
# Información general del dataset
df.info()
```

```python
# Revisar valores nulos y duplicados
print("Valores nulos:", df.isnull().sum().sum())
print("Filas duplicadas:", df.duplicated().sum())
```

```python
# Cantidad de transacciones por clase
class_counts = df["Class"].value_counts()

print("Transacciones normales:", class_counts[0])
print("Fraudes:", class_counts[1])

# Porcentaje de fraude
fraud_rate = df["Class"].mean() * 100

print(f"\nPorcentaje de fraude: {fraud_rate:.3f}%")
```

```python
# Preparar datos para la gráfica
class_data = pd.DataFrame({
    "Tipo": ["Normal", "Fraude"],
    "Cantidad": [class_counts[0], class_counts[1]]
})

# Gráfica interactiva
fig = px.bar(
    class_data,
    x="Tipo",
    y="Cantidad",
    title="Distribución de transacciones",
    text="Cantidad"
)

fig.show()
```

```python
# Crear nombres más claros para cada tipo de transacción
df["Transaction_Type"] = df["Class"].map({
    0: "Normal",
    1: "Fraude"
})

# Comparar los montos entre transacciones normales y fraudulentas
amount_summary = df.groupby("Transaction_Type")["Amount"].agg(
    Promedio="mean",
    Mediana="median",
    Minimo="min",
    Maximo="max"
).round(2)

amount_summary
```

```python
# Preparar los datos para la gráfica
amount_plot = amount_summary[["Promedio", "Mediana"]].reset_index()

# Gráfica interactiva
fig = px.bar(
    amount_plot,
    x="Transaction_Type",
    y=["Promedio", "Mediana"],
    barmode="group",
    title="Montos de transacciones: Normal vs Fraude",
    labels={
        "Transaction_Type": "Tipo de transacción",
        "value": "Monto",
        "variable": "Métrica"
    }
)

fig.show()
```

```python
# Obtener solamente las transacciones fraudulentas
fraud = df[df["Class"] == 1]

fig = px.histogram(
    fraud,
    x="Amount",
    nbins=50,
    title="Distribución de montos fraudulentos"
)

fig.show()
```

```python
# Calcular correlación usando solo columnas numéricas
correlation = df.corr(numeric_only=True)["Class"].drop("Class")

# Mostrar las 10 variables con mayor relación con fraude
top_features = correlation.abs().sort_values(ascending=False).head(10)

top_features
```
